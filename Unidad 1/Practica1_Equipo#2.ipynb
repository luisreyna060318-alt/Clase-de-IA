```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from google.colab import drive
drive.mount('/content/drive')
```

```python
forbes_df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Ejercicio 1/Forbes_2000_top_company.csv')
forbes_df
```

```python
# Gráfica de barras que muestra la frecuencia de empresas en cada industria
forbes_df['Industry'].value_counts().plot(kind='bar')
plt.title('Distribución de Empresas por Industria')
plt.xlabel('Industria')
plt.ylabel('Número de Empresas')
plt.show()
```

```python
# Gráfica de barras que muestra los 10 países con el mayor número de empresas en el ranking
forbes_df['Country'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Países con Más Empresas en el Ranking Forbes 2000')
plt.xlabel('País')
plt.ylabel('Número de Empresas')
plt.show()
```

```python
# Gráfica de barras con las 10 organizaciones con mayor valor de mercado
top10 = forbes_df.nlargest(10, 'Market Value (Billions)')
plt.bar(top10['Organization Name'], top10['Market Value (Billions)'])
plt.title('Top 10 Organizaciones por Valor de Mercado')
plt.xlabel('Organización')
plt.ylabel('Valor de Mercado (Billones)')
plt.xticks(rotation=90)
plt.show()
```

```python
# Gráfica de barras con los 10 CEOs que más veces aparecen en el ranking
forbes_df['CEO'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 CEOs con Más Apariciones en el Ranking')
plt.xlabel('CEO')
plt.ylabel('Número de Apariciones')
plt.show()
```

```python
# Boxplot para la distribución de las ganancias de las empresas
sns.boxplot(y=forbes_df['Profits (Billions)'])
plt.title('Boxplot de Ganancias (Billones)')
plt.ylabel('Ganancias (Billones)')
plt.show()
```

```python
# Boxplot de los ingresos (Revenue) de las empresas
sns.boxplot(y=forbes_df['Revenue (Billions)'])
plt.title('Boxplot de Ingresos (Billones)')
plt.xlabel('Ingresos (Billones)')
plt.show()
```

```python
# Boxplot del valor de mercado de las empresas
sns.boxplot(y=forbes_df['Market Value (Billions)'])
plt.title('Boxplot de Valor de Mercado (Billones)')
plt.xlabel('Valor de Mercado (Billones)')
plt.show()
```

```python
# Boxplot del total de empleados
sns.boxplot(x=forbes_df['Total Employees'])
plt.title('Boxplot de Total de Empleados (excluyendo datos faltantes)')
plt.xlabel('Total de Empleados')
plt.show()
```

```python
# Boxplot para la distribución de los activos de las empresas
sns.boxplot(x=forbes_df['Assets (Billions)'])
plt.title('Boxplot de Activos (Billones)')
plt.xlabel('Activos (Billones)')
plt.show()
```

```python
# Boxplot del año de fundación
sns.boxplot(x=forbes_df['Year Founded'])
plt.title('Boxplot de Año de Fundación (excluyendo datos faltantes)')
plt.xlabel('Año de Fundación')
plt.show()
```

```python
# Diagrama de dispersión de la relación entre los ingresos y las ganancias de las empresas
sns.scatterplot(x=forbes_df['Revenue (Billions)'], y=forbes_df['Profits (Billions)'])
plt.title('Ingresos vs. Ganancias (Billones)')
plt.xlabel('Ingresos (Billones)')
plt.ylabel('Ganancias (Billones)')
plt.show()
```

```python
# Diagrama de dispersión de la relación entre el valor de mercado de una empresa y su número total de empleados
sns.scatterplot(x=forbes_df['Market Value (Billions)'], y=forbes_df['Total Employees'])
plt.title('Valor de Mercado vs. Total de Empleados')
plt.xlabel('Valor de Mercado (Billones)')
plt.ylabel('Total de Empleados')
plt.show()
```

```python
#Este mapa de calor visualiza las correlaciones entre todas las variables numéricas del dataset
sns.heatmap(round(forbes_df.corr(numeric_only=True), 2), annot=True)
plt.title('Mapa de Calor entre Variables Numéricas')
plt.show()
```
